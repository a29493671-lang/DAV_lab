# Data Preprocessing - Mean Imputation

Mean imputation replaces missing numerical values with the mean of the available values.

```python
import pandas as pd

data = {
    "Name": ["Akash", "Rahul", "Priya", "Sneha", "Arjun"],
    "Age": [20, 21, None, 22, 21],
    "Marks": [85, 78, 92, None, 88]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print("Dataset after Mean Imputation:")
display(df)
```

```python
import pandas as pd

data = {
    "Name": ["Akash", "Rahul", "Priya", "Sneha", "Arjun"],
    "Marks": [85, None, 92, None, 88]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

df["Marks"] = df["Marks"].ffill()

print("Dataset after Forward Fill:")
display(df)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

data = {
    "Name": ["Akash", "Rahul", "Priya", "Sneha", "Arjun"],
    "Department": ["CSE", "ECE", "CSE", "ISE", "ECE"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

encoder = LabelEncoder()

df["Department_Encoded"] = encoder.fit_transform(df["Department"])

print("Dataset after Label Encoding:")
display(df)
```
