# Exploratory Data Analysis (EDA)

## Loading Dataset, Descriptive Statistics, Missing Values and Duplicates

This notebook demonstrates the basic steps of Exploratory Data Analysis using Pandas.

```python
import pandas as pd
```

```python
# Creating a sample student dataset

data = {
    "Name": ["Akash", "Rahul", "Priya", "Sneha", "Arjun", "Rahul", "Meena"],
    "Age": [20, 21, 20, None, 22, 21, 20],
    "Marks": [85, 78, 92, 88, None, 78, 95],
    "Department": ["CSE", "ISE", "CSE", "ECE", "CSE", "ISE", "ECE"]
}

df = pd.DataFrame(data)

df
```

```python
# Display first five rows

print("First five rows:")
display(df.head())
```

```python
# Descriptive statistics for numerical columns

print("Descriptive Statistics:")
display(df.describe())
```

```python
# Checking missing values in each column

print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Total missing values

total_missing = df.isnull().sum().sum()

print("Total missing values:", total_missing)
```

```python
# Display rows containing at least one missing value

missing_rows = df[df.isnull().any(axis=1)]

print("Rows containing missing values:")
display(missing_rows)
```

```python
# Checking for duplicate rows

duplicates = df.duplicated()

print("Duplicate rows:")
display(df[duplicates])
```

```python
# Count the number of duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```
