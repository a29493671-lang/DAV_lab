# Introduction to Python for Data Analytics

## NumPy and Pandas Basics

This notebook demonstrates:
- NumPy array creation and operations
- Array indexing and slicing
- Pandas DataFrame creation
- DataFrame indexing and slicing
- Basic data analysis using Pandas

```python
import numpy as np
import pandas as pd
```

```python
# Creating a NumPy array
marks = np.array([78, 85, 92, 67, 88])

print("Marks:", marks)
print("Array type:", type(marks))
print("Number of elements:", len(marks))
```

```python
# Basic array operations

print("Original marks:", marks)
print("Marks + 5:", marks + 5)
print("Marks - 5:", marks - 5)
print("Marks * 2:", marks * 2)
print("Marks / 2:", marks / 2)

print("Average marks:", np.mean(marks))
print("Maximum marks:", np.max(marks))
print("Minimum marks:", np.min(marks))
```

```python
# NumPy indexing starts from 0

print("First mark:", marks[0])
print("Third mark:", marks[2])
print("Last mark:", marks[-1])
```

```python
# NumPy slicing

print("First three marks:", marks[0:3])
print("Marks from index 2:", marks[2:])
print("First four marks:", marks[:4])
print("Every second mark:", marks[::2])
```

```python
# Creating a DataFrame

data = {
    "Name": ["Akash", "Rahul", "Priya", "Sneha", "Arjun"],
    "Age": [20, 21, 20, 22, 21],
    "Marks": [78, 85, 92, 67, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Selecting columns

print("Names:")
print(df["Name"])

print("\nMarks:")
print(df["Marks"])
```

```python
# Selecting rows using iloc

print("First row:")
print(df.iloc[0])

print("\nThird row:")
print(df.iloc[2])
```

```python
# Selecting multiple rows

print("First three rows:")
print(df.iloc[0:3])

print("\nRows 2 to 4:")
print(df.iloc[1:4])
```

```python
# Selecting specific rows and columns

selected_data = df.iloc[0:3, 0:2]

print(selected_data)
```

```python
# Basic analysis using Pandas

print("Average marks:", df["Marks"].mean())
print("Highest marks:", df["Marks"].max())
print("Lowest marks:", df["Marks"].min())

print("\nStudents scoring more than 80:")
print(df[df["Marks"] > 80])
```

```python
# Display the complete DataFrame

print("Student Data:")
display(df)
```
