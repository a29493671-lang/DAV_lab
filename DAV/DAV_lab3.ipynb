# Data Visualization using Matplotlib and Seaborn

## Histograms, Box Plots, Scatter Plots, Pair Plots and Correlation Heatmaps

This notebook demonstrates:
- Creating visualizations using Matplotlib and Seaborn
- Histograms for understanding data distribution
- Box plots for identifying spread and outliers
- Scatter plots for studying relationships between variables
- Pair plots for comparing multiple variables
- Correlation heatmaps for understanding relationships between numerical variables

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully")
```

```python
data = {
    "Name": ["Akash", "Rahul", "Priya", "Sneha", "Arjun",
             "Meena", "Kiran", "Anjali", "Rohit", "Pooja"],
    
    "Age": [20, 21, 20, 22, 21, 20, 23, 21, 22, 20],
    
    "StudyHours": [3, 5, 6, 2, 7, 4, 8, 6, 5, 3],
    
    "Marks": [65, 78, 85, 60, 92, 72, 95, 88, 80, 68],
    
    "Attendance": [75, 82, 90, 70, 95, 80, 98, 92, 85, 78]
}

df = pd.DataFrame(data)

print("Student Dataset:")
display(df)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Marks"], bins=5, edgecolor="black")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["Marks"])

plt.title("Box Plot of Student Marks")
plt.xlabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(x=df["StudyHours"], y=df["Marks"])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
sns.pairplot(df[["Age", "StudyHours", "Marks", "Attendance"]])

plt.show()
```

```python
correlation = df[["Age", "StudyHours", "Marks", "Attendance"]].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
print("Data Visualization Summary")
print("--------------------------")
print("1. Histogram     - Shows data distribution")
print("2. Box Plot      - Shows spread and possible outliers")
print("3. Scatter Plot  - Shows relationship between two variables")
print("4. Pair Plot     - Compares multiple numerical variables")
print("5. Heatmap       - Shows correlation between variables")
```
